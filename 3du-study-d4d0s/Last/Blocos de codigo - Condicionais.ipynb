## Condicionais


```python
# BLOCO DE CÓDIGO
if 0 == 0:
    print('é zero') # dentro de um bloco de código
```

```python
numero = 18
novaIdade = int(input('Digite sua idade: '))

if novaIdade >= numero:
    print('Pode tomar cachaça')
else:
    print('Não pode tomar cachaça')
```

```python
print('POSITIVO OU NEGATIVO\n')

num = float(input('Digite um número para saber se é POSITIVO ou NEGATIVO: '))

if num >= 0:
    print(f'{num:.0f} POSITIVO')
else:
    print(f'{num:.0f} NEGATIVO')
```
