## Tuplas

```python
numeros = (1,2,3,4,5,6,7,8,9)
print(numeros)
print(type(numeros)) # Tuplas são imutaveis
```

```python
listaSimples1 = [1,2,3,4,5]
listaSimples2 = [6,7,8,9,10]
listaTupla = (listaSimples1, listaSimples2)

print(f'Essa são as listas simples {listaSimples1} e {listaSimples2} \n Essa e lista de Tupla {listaTupla} \n ')
print(f'Tipos de listas {type(listaSimples1)} e {type(listaSimples2)}  e {type(listaTupla)}')
```
