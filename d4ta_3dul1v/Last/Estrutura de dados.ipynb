**POSIÇÃO DE UM ITEM DA LISTA**


````python
clientesLoja = [
    "McDonald's",
    "Burger King",
    "Subway",
    "Starbucks",
    "KFC",
    "Domino's Pizza",
    "Pizza Hut",
    "Havan",
    "Magazine Luiza",
    "Casas Bahia",
    "Carrefour",
    "Extra"
]

print(clientesLoja)

primeiroClinte = clientesLoja[0] # buscando o indice 0 que no caso é o primeiro a contagem de uma lista começa de 0,1,2... 
print(primeiroClinte)

clienteNcinco = clientesLoja[5]
print(clienteNcinco)

print(f'\n Os clientes que foram mias requisitados foram {primeiroClinte} e {clienteNcinco}')

procuraCliente = 'Extra'
poiscao = clientesLoja.index(procuraCliente) # index() esta buscando o nome e não o indice dentro da lista 
print(f'\n A posição do cliente {procuraCliente.upper()} dentro da nossa lista é {poiscao}')
````

# 📘 Acessando elementos em listas no Python

## 🔹 Índices com `[ ]`

Em Python, listas começam no índice **0**.

```python
lista = [10, 20, 30, 40, 50]
```

| Índice | Valor |
| ------ | ----- |
| 0      | 10    |
| 1      | 20    |
| 2      | 30    |
| 3      | 40    |
| 4      | 50    |

Exemplos:

```python
lista[0]  # 10
lista[3]  # 40
```

👉 O número dentro dos colchetes indica a posição.

---

## 🔹 Índices negativos

Você pode acessar elementos de trás pra frente:

```python
lista[-1]  # 50 (último)
lista[-2]  # 40
```

👉 Muito útil quando você quer o final da lista.

---

## 🔹 Fatiamento (slicing) com `[:]`

Formato:

```python
lista[inicio:fim]
```

⚠️ Regra:

* Inclui o índice inicial
* Exclui o índice final

Exemplo:

```python
lista = [10, 20, 30, 40, 50]

lista[1:4]  # [20, 30, 40]
```

---

## 🔹 Variações do slicing

### 1. Do começo até um ponto

```python
lista[:3]  # [10, 20, 30]
```

---

### 2. De um ponto até o final

```python
lista[2:]  # [30, 40, 50]
```

---

### 3. Copiar a lista

```python
lista[:]  # cópia da lista
```

---

### 4. Usando passo (pular elementos)

Formato completo:

```python
lista[inicio:fim:passo]
```

Exemplo:

```python
lista[::2]  # [10, 30, 50]
```

---

### 5. Inverter a lista

```python
lista[::-1]  # [50, 40, 30, 20, 10]
```

---

## 🧠 Resumo rápido

* `lista[3]` → acessa um elemento
* `lista[1:4]` → pega parte da lista
* `lista[:3]` → do início até índice 3
* `lista[2:]` → do índice 2 até o final
* `lista[::2]` → pula de 2 em 2
* `lista[::-1]` → inverte a lista

---

## 💡 Dica final

Se você dominar slicing, seu código fica:

* mais limpo
* mais rápido de escrever
* mais profissional




````python
# OBTER UMA LISTA A PARTIR DE UMA LISTA 

carros = [
    "Toyota Corolla",
    "Honda Civic",
    "Ford Mustang",
    "Chevrolet Onix",
    "Volkswagen Golf",
    "Hyundai HB20",
    "Fiat Uno",
    "Jeep Renegade",
    "BMW Série 3",
    "Audi A4"
]

print(carros)

tresPrimeiros = carros[:3]
print(tresPrimeiros)

tresPenultimos = carros[3:]
print(tresPenultimos)

tresUltimos = carros[-3:]
print(tresUltimos)
````
